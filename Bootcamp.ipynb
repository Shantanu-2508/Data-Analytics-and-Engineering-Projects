```python
correct_password = "2222"
attempts_left = 3

while attempts_left>0:
    password = input("Enter Your Password")
    if password == correct_password:
        print("Logged in successfully")
        break
        
    else:
        print("Incorrect Password")
        attempts_left = attempts_left-1
        print("There are",attempts_left,"attempts left")

if attempts_left == 0:
    print("Your account is blocked")
else:
    print("Welcome to the bank")
    
```

```python
correct_password = "2222"
attempts_left = 3

while attempts_left > 0:
    password = input("Enter Your Password: ")
    if password == correct_password:
        print("Logged in successfully")
        break # Exit the loop on success
    else:
        print("Incorrect Password")
        attempts_left -= 1 # Decrement attempts on failure
        print(f"There are {attempts_left} attempts left")

# After the loop, check the outcome
if attempts_left == 0:
    print("Your account is blocked")
else:
    print("Welcome to the bank")
```

```python
def factorial(num):
    result = 1
    if num<0:
        print("Factorial of negative number does not exist")
        
    else:
        while num>0:
            result = result*num
            num -= 1
        print(result)
```

```python
factorial(5)
```

```python
# Tax calculator

def Tax():
    salary = int(input("Enter your salary"))
    if salary <= 2500:
        print("You have to pay:",salary*0,"RS tax")
    elif salary > 2500 and salary <= 5000:
        print("You have to pay:",salary*0.05,"RS tax")
    elif salary > 5000 and salary <= 10000:
        print("You have to pay:",salary*0.10,"RS tax")
    elif salary > 10000 and salary <= 15000:
        print("You have to pay:",salary*0.20,"RS tax")
    else:
        print("You have to pay:",salary*0.30,"RS tax")
```

```python
Tax()
```
